```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

print("Libraries imported successfully!")
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
df = pd.read_csv(list(uploaded.keys())[0])

print("Dataset loaded successfully!")

df.head()
```

```python
print("First 5 Rows:")
display(df.head())

print("Dataset Shape:")
print(df.shape)

print("Column Names:")
print(df.columns)

print("Dataset Information:")
df.info()

print("Statistical Description:")
display(df.describe())
```

```python
df = df[['Age', 'Salary', 'JobSatisfaction',
         'YearsAtCompany', 'WorkLifeBalance']]

df.head()
```

```python
df = df.drop_duplicates()

df = df.dropna()

print("Cleaned Dataset:")
display(df.head())

print("Remaining Missing Values:")
print(df.isnull().sum())
```

```python
df['RiskScore'] = (
    (5 - df['JobSatisfaction']) * 10 +
    (5 - df['WorkLifeBalance']) * 10 +
    (df['Salary'].max() - df['Salary']) / 1000 +
    (20 - df['YearsAtCompany']) * 2
)

df['RiskScore'] = df['RiskScore'].clip(0, 100)

display(df.head())
```

```python
df.hist(figsize=(10, 8))

plt.suptitle("Employee Data Distribution")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(df.corr(), annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df)

plt.title("Employee Data Box Plot")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(df['Salary'], df['RiskScore'])

plt.xlabel("Salary")

plt.ylabel("Risk Score")

plt.title("Salary vs Risk Score")

plt.show()
```

```python
X = df[['Age', 'Salary', 'JobSatisfaction',
        'YearsAtCompany', 'WorkLifeBalance']]

Y = df['RiskScore']

print("Input Variables:")
display(X.head())

print("Output Variable:")
display(Y.head())
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.2,
    random_state=42
)

print("Training Data:", X_train.shape)

print("Testing Data:", X_test.shape)
```

```python
model = LinearRegression()

model.fit(X_train, Y_train)

print("Linear Regression Model Trained Successfully!")
```

```python
Y_pred = model.predict(X_test)

print("Predicted Values:")

print(Y_pred[:10])
```

```python
result = pd.DataFrame({
    'Actual': Y_test.values,
    'Predicted': Y_pred
})

display(result.head(10))
```

```python
mae = mean_absolute_error(Y_test, Y_pred)

mse = mean_squared_error(Y_test, Y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(Y_test, Y_pred)

print("Model Performance")

print("MAE:", mae)

print("MSE:", mse)

print("RMSE:", rmse)

print("R2 Score:", r2)
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(Y_test, Y_pred)

plt.xlabel("Actual Risk Score")

plt.ylabel("Predicted Risk Score")

plt.title("Actual vs Predicted Values")

plt.show()
```

```python
new_employee = pd.DataFrame({
    'Age': [25],
    'Salary': [20000],
    'JobSatisfaction': [2],
    'YearsAtCompany': [2],
    'WorkLifeBalance': [2]
})

prediction = model.predict(new_employee)

print("Predicted Risk Score:", prediction[0])
```

```python
score = prediction[0]

if score < 35:
    print("Low Risk")

elif score < 65:
    print("Medium Risk")

else:
    print("High Risk")
```
